```python
boolean_values = [True, False]
kb_models = []

print("i) FULL TRUTH TABLE")
print("-" * 55)
print(f"{'P':<6} {'Q':<6} {'R':<6} | {'Q->P':<6} {'P->~Q':<6} {'Q v R':<6} | {'KB':<6}")
print("-" * 55)

for P in boolean_values:
    for Q in boolean_values:
        for R in boolean_values:
            q_implies_p = (not Q) or P
            p_implies_not_q = (not P) or (not Q)
            q_or_r = Q or R
            
            KB = q_implies_p and p_implies_not_q and q_or_r
            
            print(f"{str(P):<6} {str(Q):<6} {str(R):<6} | "
                  f"{str(q_implies_p):<6} {str(p_implies_not_q):<6} {str(q_or_r):<6} | "
                  f"{str(KB):<6}")
            if KB:
                kb_models.append({'P': P, 'Q': Q, 'R': R})

print("\n" + "_"*55)
print("ii) Evaluation Table for: Does KB entail R?")
print("_"*55)
print(f"{'Model Row':<12} | {'P':<6} {'Q':<6} {'R':<6} | {'KB':<6} | {'Query (R)':<10}")
print("-" * 55)

entail_ii = True
for idx, model in enumerate(kb_models):
    query_val = model['R']
    if not query_val:
        entail_ii = False
    print(f"KB Model {idx+1:<3} | {str(model['P']):<6} {str(model['Q']):<6} {str(model['R']):<6} | {'True':<6} | {str(query_val):<10}")

print("-" * 55)
print(f"Conclusion: Does KB entail R? -> {entail_ii}\n")

print("_"*55)
print("iii) Evaluation Table for: Does KB entail R -> P?")
print("_"*55)
print(f"{'Model Row':<12} | {'P':<6} {'Q':<6} {'R':<6} | {'KB':<6} | {'Query (R->P)':<12}")
print("-" * 55)

entail_iii = True
for idx, model in enumerate(kb_models):
    query_val = (not model['R']) or model['P']
    if not query_val:
        entail_iii = False
    print(f"KB Model {idx+1:<3} | {str(model['P']):<6} {str(model['Q']):<6} {str(model['R']):<6} | {'True':<6} | {str(query_val):<12}")

print("-" * 55)
print(f"Conclusion: Does KB entail R -> P? -> {entail_iii}\n")

print("_"*55)
print("iv) Evaluation Table for: Does KB entail Q -> R?")
print("_"*55)
print(f"{'Model Row':<12} | {'P':<6} {'Q':<6} {'R':<6} | {'KB':<6} | {'Query (Q->R)':<12}")
print("-" * 55)

entail_iv = True
for idx, model in enumerate(kb_models):
    query_val = (not model['Q']) or model['R']
    if not query_val:
        entail_iv = False
    print(f"KB Model {idx+1:<3} | {str(model['P']):<6} {str(model['Q']):<6} {str(model['R']):<6} | {'True':<6} | {str(query_val):<12}")

print("-" * 55)
print(f"Conclusion: Does KB entail Q -> R? -> {entail_iv}\n")
```
